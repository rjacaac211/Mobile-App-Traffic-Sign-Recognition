```python
import os
import shutil
import random
import csv
```

```python
# Paths setup
dataset_dir = 'tsc_dataset/crop'
train_dir = 'Train'
test_dir = 'Test'
csv_filename = 'test_labels.csv'
```

```python
# Make sure train and test directories exist
os.makedirs(train_dir, exist_ok=True)
os.makedirs(test_dir, exist_ok=True)
```

```python
# Parameters
test_split = 0.2  # Percentage of images to be moved to the test set

# CSV file setup
with open(csv_filename, 'w', newline='') as file:
    writer = csv.writer(file)
    writer.writerow(['ClassId', 'Path'])

    # Loop through each class directory in the dataset
    for class_id in os.listdir(dataset_dir):
        class_dir = os.path.join(dataset_dir, class_id)
        images = os.listdir(class_dir)
        
        # Ensure subdirectory for class exists in train_dir
        class_train_dir = os.path.join(train_dir, class_id)
        os.makedirs(class_train_dir, exist_ok=True)
        
        # Determine train and test split
        random.shuffle(images)
        split_point = int(len(images) * test_split)
        test_images = images[:split_point]
        train_images = images[split_point:]

        # Copy images to the respective 'Train' and 'Test' directories
        for image in test_images:
            src = os.path.join(class_dir, image)
            dst = os.path.join(test_dir, f"{class_id}_{image}")
            shutil.copy(src, dst)
            writer.writerow([class_id, os.path.join('Test', f"{class_id}_{image}")])

        for image in train_images:
            src = os.path.join(class_dir, image)
            dst = os.path.join(class_train_dir, image)
            shutil.copy(src, dst)
```
